# Week 1 — Keyword Search Baseline

**Project:** BC Federal Labor Jurisdiction Engine

**Purpose:** Build a simple, reproducible baseline that imitates a traditional `Ctrl + F` exact keyword search.  
This baseline demonstrates why literal matching is insufficient when federal and provincial documents discuss overlapping topics such as sick leave and overtime.


```python
from typing import List

def local_keyword_matcher(keyword: str, document_text: str, top_n: int = 3) -> List[str]:
    """Return up to top_n lines containing the exact keyword string, case-insensitively."""
    if not isinstance(keyword, str) or not keyword.strip():
        raise ValueError("keyword must be a non-empty string")
    if not isinstance(document_text, str):
        raise TypeError("document_text must be a string")
    if top_n < 1:
        raise ValueError("top_n must be at least 1")

    lines = document_text.splitlines()
    matches = [
        line.strip()
        for line in lines
        if keyword.lower() in line.lower() and line.strip()
    ]
    return matches[:top_n]
```

```python
# Simulated overlapping legal text snippets
sample_jurisdiction_data = """
[BC Provincial Law] ESA Section 49.1: Eligible employees get 5 paid sick days per calendar year after meeting the eligibility requirement.
[Federal Canada Law] CLC Section 239: Federally regulated employees can earn up to 10 paid medical leave days.
[BC Provincial Law] ESA Part 4 Section 40: Overtime is generally calculated at 1.5x regular pay after 8 hours in a workday.
[Federal Canada Law] CLC Part III Section 174: Overtime is paid at 1.5x the regular rate for hours worked beyond standard limits.
"""

test_keyword = "paid"
baseline_output = local_keyword_matcher(test_keyword, sample_jurisdiction_data)

print(f"--- Raw Keyword Match Results for: '{test_keyword}' ---")
for index, line_match in enumerate(baseline_output, 1):
    print(f"{index}. {line_match}")
```

## Expected interpretation

The baseline returns literal matches from both legal frameworks. It does **not** understand:

- whether the worker is federally or provincially regulated;
- whether the retrieved rule applies to the user's employer or industry;
- whether two similar phrases have different legal meanings;
- which source should be prioritized.

This limitation motivates the Week 2 context-aware vector retrieval pipeline.


```python
# Optional: test several terms and compare the mixed results
for keyword in ["sick", "overtime", "1.5x"]:
    print(f"\nKeyword: {keyword}")
    results = local_keyword_matcher(keyword, sample_jurisdiction_data, top_n=5)
    if not results:
        print("  No exact matches")
    else:
        for i, result in enumerate(results, 1):
            print(f"  {i}. {result}")
```

# Week 1 Baseline Result

The keyword-search baseline successfully returned exact text matches from the simulated legal dataset. However, the results demonstrate an important limitation of traditional keyword retrieval. When the terms “overtime” and “1.5x” were searched, the system returned both British Columbia provincial rules and federal Canada Labour Code rules.

Although the matched sentences were technically relevant to the keywords, the baseline could not determine which jurisdiction applied to the user’s employment situation. It also could not prioritize the correct source based on the worker’s employer, industry, or location. This result confirms that exact keyword matching alone is insufficient for a labour jurisdiction assistant and supports the need for the semantic vector retrieval pipeline developed in Week 2.
